**Requests** allows you to send HTTP/1.1 requests extremely easily. There’s no need to manually add query strings to your URLs, or to form-encode your POST data. Keep-alive and HTTP connection pooling are 100% automatic, thanks to urllib3.

image.png

```python
import requests

#  URL to scrape
url = 'http://www.webscrapingfordatascience.com/basichttp/'

# Send GET request
r = requests.get(url)

# Print the raw HTML response
print(r.text)
```

- **HTTP GET** action: the **requests.get** method is used to perform a GET request to the provided URL.
- Requests will make sure to format a proper HTTP request message.
- Requests takes care of parsing the HTTP reply so that you can immediately start working with it.
- **r.text** contains the HTTP response content body in a textual form.

# Lets see more details

```python
import requests

# Define the URL to scrape
url = 'http://www.webscrapingfordatascience.com/basichttp/'

# Send the GET request
r = requests.get(url)

## r.status_code & r.reason: How the server responded.
# Print status code (e.g., 200)
print(r.status_code)

# Print status message (e.g., OK)
print(r.reason)


### r.headers: What the server sent back.
# Print HTTP response headers
print(r.headers)

# Print request object
print(r.request)


## r.request.headers: What headers our script sent to the server.
# Print request headers (sent by our scraper)
print(r.request.headers)


## r.text: The web page content.
# Print the actual HTML/text content of the page
print(r.text)
```

## ANother example: find by tag

```python
# Step 1: Import required libraries
import requests
from bs4 import BeautifulSoup

# Step 2: Define the target URL (Game of Thrones Wikipedia page)
url = 'https://en.wikipedia.org/w/index.php?title=List_of_Game_of_Thrones_episodes&oldid=802553687'

# Step 3: Send an HTTP GET request
r = requests.get(url)
html_contents = r.text  # Get the HTML content as plain text

# Step 4: Parse the HTML using BeautifulSoup
html_soup = BeautifulSoup(html_contents, 'html.parser')

# Step 5: Find and print the first <h1> tag
print(html_soup.find('h1'))


```

## Example: find by 'id'

```python
# Step 1: Import required libraries
import requests
from bs4 import BeautifulSoup

# Step 2: Define the target URL (Game of Thrones Wikipedia page)
url = 'https://en.wikipedia.org/w/index.php?title=List_of_Game_of_Thrones_episodes&oldid=802553687'

# Step 3: Send an HTTP GET request
r = requests.get(url)
html_contents = r.text  # Get the HTML content as plain text

# Step 4: Parse the HTML using BeautifulSoup
html_soup = BeautifulSoup(html_contents, 'html.parser')

# Step 5:  Find an element with id = 'Episodes'

print(html_soup.find(id='Episodes'))
```
